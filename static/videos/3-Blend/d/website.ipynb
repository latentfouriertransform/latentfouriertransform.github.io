# 3 - Blending

```python
HERO_STYLE["legend_loc"] = "center right"
HERO_STYLE["figsize"] = (12, 7)

letters = ["a", "b", "c", "d"]

in_paths =[
    "/data/hai-res/ycda/gen/notebooks/9-3/Blending_1/Isolate_then_Blend.wav",
    "/data/hai-res/ycda/gen/notebooks/9-3/Blending_2/Isolate_then_Blend_2.wav",
    "/data/hai-res/ycda/gen/notebooks/9-3/Blending_3/Isolate_then_Blend_3a.wav",
    "/data/hai-res/ycda/gen/notebooks/9-29-Demos/3-Blend/d/storage/compiled.pt"   
]

all_low_highs_Hz_1 = [
    [3.5, 4.5],
    [0, 1],
    [0, 0.5],
    [7.5, 8],
]
all_low_highs_Hz_2 = [
    [0, 1],
    [12, 18],
    [20, 40],
    [0, 0.25],
]

all_axes = [LATENT_AXES, LATENT_AXES_FULL, LATENT_AXES_FULL, LATENT_AXES]

def generate_blend_videos(
    letter="",
    low_highs_Hz_1=[],
    low_highs_Hz_2=[],
    in_path="",
    axes={},
):
    
    save_names = [f"3{letter}_input_1", f"3{letter}_iso_1", f"3{letter}_input_2", f"3{letter}_iso_2", f"3{letter}_blend"]

    base_dir = "/data/hai-res/ycda/gen/notebooks/9-29-Demos/3-Blend/" + letter
    titles=["Input 1",
            f"Input 1 Isolated to {low_highs_Hz_1[0]}-{low_highs_Hz_1[1]} Hz", 
            "Input 2",
            f"Input 2 Isolated to {low_highs_Hz_2[0]}-{low_highs_Hz_2[1]} Hz",
            "Blended"
    ]
    
    low_high_1_all = [
        [0, latent_nyq],
        low_highs_Hz_1,
        low_highs_Hz_1,
        low_highs_Hz_1,
        low_highs_Hz_1,
    ]
    low_high_2_all = [
        [0, latent_nyq],
        [0, latent_nyq],
        [0, latent_nyq],
        low_highs_Hz_2,
        low_highs_Hz_2,
    ]
    floor_ceilings_1 = [
        [0, 1],
        [0, 1],
        [0, 0],
        [0, 0],
        [0, 1],
    ]
    floor_ceilings_2 = [
        [0, 0],
        [0, 0],
        [0, 1],
        [0, 1],
        [0, 1],
    ]
    
    for i in range(5):
        track_1 = Track(
            "Input 1",
            low_highs=[low_high_1_all[i]],
            floor_ceilings=[floor_ceilings_1[i]],
            band_transition_width=0.001
        )
        
        track_2 = Track(
            "Input 2",
            low_highs=[low_high_2_all[i]],
            floor_ceilings=[floor_ceilings_2[i]],
            color="#45FF17",
            band_transition_width=0.001,
        )

        if i < 2:
            tracks = [track_1]
        elif i < 4:
            tracks = [track_2]
        else:
            tracks = [track_1, track_2]


        
        anim = BandpassAnimation(tracks=tracks, titles=[titles[i]], style=HERO_STYLE, axes=axes)

        silent_path = os.path.join(base_dir, f"{save_names[i]}_silent.mp4")

        anim.animate((131072 / 22050), save_path=silent_path)
        # Save Audio
        print(in_path)
        
        if in_path[-4:] == ".wav":
            all_audios = torchaudio.load(in_path)[0].reshape(5, 131072)
        else:
            all_audios = torch.load(in_path).reshape(5, 131072)

        wav_out_path = os.path.join(base_dir, f"{save_names[i]}.wav")
        
        torchaudio.save(wav_out_path, all_audios[i].reshape(1,-1), sample_rate=22050)
        
        # Mux
        muxed_path = os.path.join(base_dir, f"{save_names[i]}.mp4")
        mux(silent_path, wav_out_path, muxed_path)
        
        os.remove(silent_path)
        os.remove(wav_out_path)



# 2b
# Get Audio FIles
for i in range(3,4):
    generate_blend_videos(
        letters[i],
        all_low_highs_Hz_1[i],
        all_low_highs_Hz_2[i],
        in_paths[i],
        all_axes[i],
    )
```
